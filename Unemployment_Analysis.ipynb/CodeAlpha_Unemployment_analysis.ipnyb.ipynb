```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
df = pd.read_excel("Unemployment in India.xlsx")
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Remove rows with missing values
df.dropna(inplace=True)

# Remove duplicate records
df.drop_duplicates(inplace=True)

# Reset index
df.reset_index(drop=True, inplace=True)

# Check cleaned dataset
print("Cleaned Dataset Shape:", df.shape)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())
```

```python
# Statistical Summary
print("Statistical Summary of Unemployment Dataset:")
df.describe()
```

```python
# Average Unemployment Rate by State

state_unemployment = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean()

print("Average Unemployment Rate by State:")
print(state_unemployment.sort_values(ascending=False))
```

```python
# State-wise Unemployment Rate Bar Chart

plt.figure(figsize=(14, 7))

state_unemployment.sort_values().plot(
    kind='bar',
    color='skyblue'
)

plt.title('Average Unemployment Rate by State in India')
plt.xlabel('States')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Convert Date column into datetime
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# Calculate monthly average unemployment rate
monthly_unemployment = df.groupby('Date')['Estimated Unemployment Rate (%)'].mean()

# Plot the trend
plt.figure(figsize=(12, 6))

plt.plot(monthly_unemployment.index,
         monthly_unemployment.values,
         marker='o',
         linewidth=2)

plt.title('Monthly Unemployment Rate Trend in India')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Create period categories
df['Period'] = df['Date'].apply(
    lambda x: 'Before COVID' if x < pd.Timestamp('2020-03-01') else 'During COVID'
)

# Calculate average unemployment rate
covid_comparison = df.groupby('Period')['Estimated Unemployment Rate (%)'].mean()

print("Unemployment Rate Comparison:")
print(covid_comparison)
```

```python
plt.figure(figsize=(8, 5))

covid_comparison.plot(
    kind='bar',
    color=['skyblue', 'salmon']
)

plt.title('Impact of COVID-19 on Unemployment Rate in India')
plt.xlabel('Period')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Calculate average unemployment rate by area
area_unemployment = df.groupby('Area')['Estimated Unemployment Rate (%)'].mean()

print("Rural vs Urban Unemployment Rate:")
print(area_unemployment)
```

```python
plt.figure(figsize=(8, 5))

area_unemployment.plot(
    kind='bar',
    color=['lightgreen', 'orange']
)

plt.title('Unemployment Rate: Rural vs Urban Areas')
plt.xlabel('Area')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# State with highest unemployment rate
highest_state = state_unemployment.idxmax()
highest_rate = state_unemployment.max()

# State with lowest unemployment rate
lowest_state = state_unemployment.idxmin()
lowest_rate = state_unemployment.min()

print("Highest Unemployment Rate:")
print(highest_state, ":", round(highest_rate, 2), "%")

print("\nLowest Unemployment Rate:")
print(lowest_state, ":", round(lowest_rate, 2), "%")
```

```python
top_10_states = state_unemployment.sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(12, 6))

top_10_states.plot(
    kind='bar',
    color='coral'
)

plt.title('Top 10 States with Highest Unemployment Rate')
plt.xlabel('States')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print("UNEMPLOYMENT ANALYSIS IN INDIA")
print("--------------------------------")

print("Total Records:", len(df))
print("Total States/Regions:", df['Region'].nunique())

print("\nHighest Unemployment State:")
print(highest_state, "-", round(highest_rate, 2), "%")

print("\nLowest Unemployment State:")
print(lowest_state, "-", round(lowest_rate, 2), "%")

print("\nRural vs Urban Analysis:")
print(area_unemployment)

print("\nCOVID Impact Analysis:")
print(covid_comparison)

print("\nProject Completed Successfully!")
```
